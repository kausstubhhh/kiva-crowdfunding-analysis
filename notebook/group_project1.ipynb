### Processing Modules and Algorithms

__Data Cleaning & Standardization__
* Purpose: Correct inconsistencies, handle missing values, standardize text fields, and align country names with MPI data.

__Interactive Dashboard__
* Purpose: Provide real-time filtering, visualization, and model prediction exploration for users.
* Techniques: IPython widgets for selection, Plotly for dynamic charts, on-the-fly predictions using trained models.

#### _Brief Explanation of cell 3_

_This cell defines a set of utility functions to clean and standardize the dataset. The first two helpers safely convert selected columns to numeric or datetime formats, coercing invalid values to NaN to prevent errors during analysis. The gender_bucket_val function processes the borrower gender field and categorizes each entry into standard groups (female, male, mixed, or unknown) ensuring consistent gender classification for cleaner analysis._

```python
# Utility functions
def safenum(df, cols):
    for c in cols: df[c] = pd.to_numeric(df[c], errors='coerce')

def safedatetime(df, cols):
    for c in cols: df[c] = pd.to_datetime(df[c], errors='coerce')

def gender_bucket_val(s):
    if pd.isna(s) or str(s).strip()=='' : return 'unknown'
    s = str(s).lower()
    genders = set([g.strip() for g in s.replace("|",",").split(',') if g.strip()])
    if genders=={'female'}: return 'female'
    if genders=={'male'}: return 'male'
    if 'female' in genders and 'male' in genders: return 'mixed'
    return 'unknown'
```

#### _Brief Explanation of cell 4_

_This cell defines a compact function for data loading and cleaning. It reads the raw Kiva loan data and MPI regional data, converts important columns to numeric and datetime formats, and removes rows with missing or invalid fields. It standardizes country names, aligns each loan with its corresponding MPI region, normalizes categorical fields, and filters out inconsistent values. At the end the function selects the final cleaned feature set, converts key fields to categorical types, and optionally saves the cleaned dataset for further uses. The returned DataFrame serves as the foundation for all subsequent analysis and modeling._

```python
# Load & clean
def Cleaner_jannu(data_path=DATA, save_clean=True):
    loans = pd.read_csv(data_path / 'kiva_loans.csv', usecols=[
        'id','funded_amount','loan_amount','activity','sector','use','country_code',
        'country','region','partner_id','posted_time','disbursed_time','funded_time',
        'term_in_months','lender_count','borrower_genders'
    ])
    mpi = pd.read_csv(data_path / 'kiva_mpi_region_locations.csv')

    # Numeric & datetime coercion
    safenum(loans, ['loan_amount','funded_amount','term_in_months','lender_count'])
    safedatetime(loans, ['posted_time','disbursed_time','funded_time'])

    # Basic dropna and country clean
    loans.dropna(subset=['loan_amount','funded_amount','posted_time'], inplace=True)
    loans['country'] = loans['country'].astype(str).str.strip()
    loans = loans[loans['country'].str.match(r'^[A-Za-z\s]+$')]

    valid_countries = mpi['country'].unique()
    loans = loans[loans['country'].isin(valid_countries)]
    country_region_map = mpi.set_index('country')['region'].to_dict()
    loans['region'] = loans['country'].map(country_region_map)

    # Trim/normalize categorical columns and drop rows with missing critical categories
    for col in ['sector','activity','borrower_genders','partner_id']:
        loans[col] = loans[col].astype(str).str.strip().replace({'nan':'', '':np.nan})
    loans.dropna(subset=['sector','activity','borrower_genders','partner_id'], inplace=True)

    # Final selection and types
    clean_cols = ['id','loan_amount','funded_amount','activity','sector','borrower_genders',
                  'country','region','term_in_months','lender_count','posted_time','funded_time','partner_id']
    loans = loans[clean_cols].copy()
    for c in ['country','region','sector','activity','borrower_genders','partner_id']:
        loans[c] = loans[c].astype('category')

    if save_clean:
        loans.to_csv(data_path / 'kiva_loans_clean_final.csv', index=False)
    return loans
```

##### The cleaned dataset contains only valid loans with standardized country names, numeric/date conversions, and critical fields filled. This forms the base for all analysis.

#### _Brief Explanation of cell 5_

_This cell adds a function that upgrades to the cleaned loan dataset with thematic information. It loads the loan-theme mapping file, and combines with the current loan data, and then groups records then groups the records by loan ID. Numerical and categorical fields are kept for each loan, and all related themes are captured by combining theme-related columns into semicolon-separated strings. The enriched dataset, which offers a more comprehensive depiction of each loan's contextual themes, is then stored for future use._


```python
# Enrich with themes (keeps code compact)
def enricher_jannu(loans, data_path=DATA):
    t = pd.read_csv(data_path / 'loan_theme_ids.csv', usecols=['id','Loan Theme ID','Loan Theme Type','Partner ID'])
    merged = loans.merge(t, on='id', how='left')
    grouped = merged.groupby('id').agg({
        'loan_amount':'first','funded_amount':'first','activity':'first','sector':'first',
        'borrower_genders':'first','country':'first','region':'first','term_in_months':'first',
        'lender_count':'first','posted_time':'first','funded_time':'first','partner_id':'first',
        'Loan Theme ID': lambda x: ';'.join(x.dropna().astype(str)),
        'Loan Theme Type': lambda x: ';'.join(x.dropna().astype(str))
    }).reset_index()
    grouped.to_csv(DATA / 'kiva_loans_clean_with_themes.csv', index=False)
    return grouped
```

##### Consolidated loan themes for every loan are now included in the enriched dataset, allowing for thematic analysis without duplication or missing values.

#### _Brief Explanation of cell 6_

_Here we have defined a function that does feature engineering inorder to prepare our dataset for analysis and modeling. Missing values are assinged to "unknown" category by ensuring that important categorical fields are standardized. To show whether the entire loan amount was successfully funded, a new binary target variable called funded is created.  The function then calculates the funding duration in days and, for robustness, substitutes the median valid value for any missing or invalid durations.  Lastly, it uses the previous utility function to create a simplified gender grouping (gender_bucket).  All of the engineered features required for downstream exploration and machine-learning tasks are present in the final DataFrame._



```python
# Feature engineering + summaries
def prepare_features(loans):
    loans = loans.copy()
    loans['borrower_genders'] = loans['borrower_genders'].astype(str).replace('nan','unknown')
    loans['region'] = loans['region'].cat.add_categories('unknown').fillna('unknown')
    loans['partner_id'] = loans['partner_id'].cat.add_categories('unknown').fillna('unknown')
    loans['funded'] = (loans['funded_amount'] >= loans['loan_amount']).astype(int)

    # funding duration with median imputation for non-positive/missing
    loans['funding_duration_days'] = (loans['funded_time'] - loans['posted_time']).dt.days
    med = loans.loc[loans['funding_duration_days']>0,'funding_duration_days'].median()
    loans['funding_duration_days'] = loans['funding_duration_days'].apply(lambda x: med if (pd.isna(x) or x<=0) else x)

    loans['gender_bucket'] = loans['borrower_genders'].apply(gender_bucket_val).astype('category')
    return loans
```

##### The output includes new features like funded (binary target), funding_duration_days, and gender_bucket, preparing the data for both EDA and modeling.

#### _Brief Explanation of cell 7_

_A function for exploratory data analysis (EDA) and visualisations is defined in this cell.  After creating numerical summaries and saving them to CSV, it computes statistics by sector and country, such as loan counts, total amounts, funding rates, and median funding durations.  Plotly is used to create interactive visualisations, such as a line chart displaying average monthly funding durations, a treemap for sector distributions, and a choropleth map for loans by country.  To find connections between numerical features, a sample correlation heatmap is also plotted using Seaborn.  Lastly, a stacked bar chart provides a brief summary of the dataset's major patterns and trends by displaying funding rates across sectors divided by borrower gender._

```python
# EDA plots (plotly + matplotlib where necessary)
def eda_saver(loans):
    # numeric summary
    loans[['loan_amount','funded_amount','term_in_months','lender_count','funding_duration_days']].describe().T.to_csv(OUT/'numeric_summary.csv')

    # country & sector stats
    country_stats = loans.groupby('country', observed=True).agg(
        num_loans=('id','count'), total_amount=('loan_amount','sum'), funding_rate=('funded','mean'), median_duration=('funding_duration_days','median')
    ).sort_values('num_loans', ascending=False).reset_index()
    country_stats.to_csv(OUT/'country_stats_full.csv', index=False)

    sector_stats = loans.groupby('sector', observed=True).agg(
        num_loans=('id','count'), total_amount=('loan_amount','sum'), funding_rate=('funded','mean'), median_duration=('funding_duration_days','median')
    ).sort_values('num_loans', ascending=False).reset_index()
    sector_stats.to_csv(OUT/'sector_stats_full.csv', index=False)

    # interactive figures
    px.choropleth(country_stats, locations='country', locationmode='country names', color='num_loans', hover_data=['total_amount','funding_rate'], title='Number of Kiva Loans by Country').update_layout(height=600).show()
    px.treemap(sector_stats, path=['sector'], values='num_loans', color='total_amount', color_continuous_scale='Viridis', title='Sector-wise Distribution of Kiva Loans').update_layout(height=500).show()

    # monthly
    loans['post_month'] = loans['posted_time'].dt.to_period('M').dt.to_timestamp()
    monthly = loans.groupby('post_month').agg(num_loans=('id','count'), avg_funding_duration=('funding_duration_days','mean'), funding_rate=('funded','mean')).reset_index()
    px.line(monthly, x='post_month', y='avg_funding_duration', title='Average Funding Duration (days) — Monthly').update_layout(height=400).show()

    # correlation sample heatmap
    num_cols = ['loan_amount','funded_amount','term_in_months','lender_count','funding_duration_days']
    sample = loans[num_cols].sample(n=min(20000,len(loans)), random_state=42)
    corr = sample.corr()
    plt.figure(figsize=(7,6)); sns.heatmap(corr, annot=True, fmt='.2f'); plt.title('Correlation Matrix (sample)'); plt.tight_layout(); plt.show()

    # sector x gender stacked funding rate
    pivot = loans.pivot_table(index='sector', columns='gender_bucket', values='funded', aggfunc='mean').fillna(0)
    for g in ['female','male','mixed','unknown']:
        if g not in pivot.columns: pivot[g]=0
    pivot = pivot[['female','male','mixed','unknown']]
    px.bar(pivot, x=pivot.index, y=pivot.columns.tolist(), title='Funding Rate by Sector & Gender', labels={'value':'Funding Rate','sector':'Sector'}, text_auto='.2f').update_layout(height=400, xaxis_tickangle=-45).show()
```

#### _Brief Explanation of cell 8_

_This cell defines a compact modeling function that trains both a classification and a regression model on the loan dataset.  It divides the data into training and test sets after preparing categorical and numerical features with robust scaling and one-hot encoding.  Loan funding success is predicted by a RandomForestClassifier, whose performance is assessed using AUC and displayed as a ROC curve. The top predictors' feature importances are also plotted.  The most important features are shown in a bar chart, and RMSE is used to measure the accuracy of a RandomForestRegressor funding duration prediction in days.  For future use, the function returns the preprocessed objects and trained models._

##### _Understanding what is AUC score and ROC curve_
_So AUC score measures how well a classification model (Randomforest in our case) can seperate two classes. This depends on the AUC value we get based on our model seperation capability._

_Interpretation:_
| AUC Value       | Meaning                            |
| --------------- | ---------------------------------- |
| **0.90 – 1.00** | Excellent model                    |
| **0.80 – 0.90** | Very good model                    |
| **0.70 – 0.80** | Good/acceptable                    |
| **0.60 – 0.70** | Poor                               |
| **0.50**        | No skill (same as random guessing) |

_ROC stands for Receiver Operating Characteristic Curve. It tells how the model performs across different classification thresholds._
###### _Y-axis(TPR):True Positive Rate(Recall);_
###### _X-axis(FPR): False Positive Rate_
_The ROC curve tells you how well the model distinguishes between the two classes._
###### _A good ROC curve:_
_Bends toward the top-left corner_
_Shows high TPR and low FPR_
###### _A bad ROC curve:_
_Lies close to the diagonal line (random guessing)_



```python
# Modeling: training + diagnostics
def train_models(loans, random_state=42):
    features_cat = ['country','region','sector','activity','partner_id','gender_bucket']
    features_num = ['loan_amount','term_in_months','lender_count']
    X = loans[features_cat + features_num]
    y_clf = loans['funded']
    y_reg = loans['funding_duration_days']

    # Using train_test_split we split the data into training and testing sets
    X_train, X_test, y_train_clf, y_test_clf, y_train_reg, y_test_reg = train_test_split(
        X, y_clf, y_reg, test_size=0.2, random_state=random_state)

    # Encodes categories and scales numeric features (OneHotEncoder, RobustScaler)
    ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False) if hasattr(OneHotEncoder(),'sparse_output') else OneHotEncoder(handle_unknown='ignore', sparse=False)
    X_train_cat = ohe.fit_transform(X_train[features_cat]); X_test_cat = ohe.transform(X_test[features_cat])
    scaler = RobustScaler()
    X_train_num = scaler.fit_transform(X_train[features_num]); X_test_num = scaler.transform(X_test[features_num])
    X_train_final = np.hstack([X_train_num, X_train_cat])
    X_test_final = np.hstack([X_test_num, X_test_cat])

    # Random Forest models are used for classification and regression
    clf = RandomForestClassifier(n_estimators=80, max_depth=12, random_state=random_state, n_jobs=-1)
    clf.fit(X_train_final, y_train_clf)
    y_prob = clf.predict_proba(X_test_final)[:,1]
    # Computes AUC score and ROC curve for classification performance
    auc = roc_auc_score(y_test_clf, y_prob)
    print('Classification AUC (Funding Success):', round(auc,4))

    fpr,tpr,_ = roc_curve(y_test_clf, y_prob)
    plt.figure(figsize=(6,5)); plt.plot(fpr,tpr,label=f'AUC={auc:.3f}'); plt.plot([0,1],[0,1],'--',color='gray'); plt.xlabel('FPR'); plt.ylabel('TPR'); plt.title('ROC Curve — Funding Success'); plt.legend(); plt.show()

    feat_names = features_num + list(ohe.get_feature_names_out(features_cat))
    feat_imp_clf = pd.DataFrame({'feature':feat_names,'importance':clf.feature_importances_}).sort_values('importance',ascending=False)
    plt.figure(figsize=(10,6)); sns.barplot(x='importance', y='feature', data=feat_imp_clf.head(10)); plt.title('Top Features — Funding Success'); plt.tight_layout(); plt.show()

    reg = RandomForestRegressor(n_estimators=80, max_depth=12, random_state=random_state, n_jobs=-1)
    reg.fit(X_train_final, y_train_reg)
    y_pred_reg = reg.predict(X_test_final)
    rmse = np.sqrt(np.mean((y_test_reg - y_pred_reg)**2))
    print('Regression RMSE (Funding Duration Days):', round(rmse,2))

    feat_imp_reg = pd.DataFrame({'feature':feat_names,'importance':reg.feature_importances_}).sort_values('importance',ascending=False)
    plt.figure(figsize=(10,6)); sns.barplot(x='importance', y='feature', data=feat_imp_reg.head(20)); plt.title('Top Features — Funding Duration'); plt.tight_layout(); plt.show()

    return {'clf':clf,'reg':reg,'ohe':ohe,'scaler':scaler}
```

Key summary metrics, funding rates by sector and gender, and model-based forecasts for funding duration and probability are also shown on the dashboard.  The dashboard remains responsive and error-free under all filtering conditions thanks to built-in guards that handle missing data or absent models._

```python
# Fully safe interactive dashboard
def interactive_dashboard(loans, models):
    loans = loans.copy()
    loans['gender_bucket'] = loans['borrower_genders'].apply(gender_bucket_val).astype('category')
    
    # Fill missing categories for safe filtering
    for c in ['region','partner_id']:
        if c in loans.columns:
            try:
                loans[c] = loans[c].cat.add_categories('unknown')
            except: 
                pass
            loans[c] = loans[c].fillna('unknown')

    countries = sorted(loans['country'].astype(str).unique())
    sectors = sorted(loans['sector'].astype(str).unique())
    genders = ['female','male','mixed','unknown']

    def notebook_dashboard(country_filter=[], sector_filter=[], gender_filter=[]):
        df = loans.copy()
        if country_filter: df = df[df['country'].isin(country_filter)]
        if sector_filter: df = df[df['sector'].isin(sector_filter)]
        if gender_filter: df = df[df['gender_bucket'].isin(gender_filter)]
        
        if df.empty:
            display(HTML("<b>No data available for this filter selection.</b>"))
            return
        
        # Summary metrics
        total = len(df)
        funded_cnt = df['funded'].sum()
        avg_loan = df['loan_amount'].mean()
        avg_dur = df['funding_duration_days'].mean()
        display(HTML(f"<b>Total loans:</b> {total:,} &nbsp;&nbsp; "
                     f"<b>Fully funded:</b> {funded_cnt:,} &nbsp;&nbsp; "
                     f"<b>Avg loan:</b> ${avg_loan:,.0f} &nbsp;&nbsp; "
                     f"<b>Avg duration (days):</b> {avg_dur:.1f}"))

        # Country map
        country_summary = df.groupby('country', observed=True).agg(num_loans=('id','count')).reset_index()
        if not country_summary.empty and country_summary['num_loans'].sum() > 0:
            px.choropleth(country_summary, locations='country', locationmode='country names', color='num_loans',
                          title='Loans by Country').update_layout(height=400).show()
        else:
            display(HTML("<b>No country map data to display.</b>"))

        # Sector treemap — safe for single-sector or zero-sector cases
        sector_summary = df.groupby('sector', observed=True).agg(
            num_loans=('id','count'), total_amount=('loan_amount','sum')).reset_index()
        if len(sector_summary) > 0 and sector_summary['total_amount'].sum() > 0:
            px.treemap(sector_summary, path=['sector'], values='num_loans', color='total_amount', 
                       title='Sector Distribution').update_layout(height=400).show()
        else:
            display(HTML("<b>Treemap skipped: not enough sector data to display.</b>"))

        # Gender bar
        gender_summary = df.groupby('gender_bucket', observed=True).agg(
            num_loans=('id','count'), funding_rate=('funded','mean')).reset_index()
        if not gender_summary.empty and gender_summary['num_loans'].sum() > 0:
            px.bar(gender_summary, x='gender_bucket', y='funding_rate', text='num_loans', 
                   title='Funding Rate by Gender').update_traces(textposition='outside').update_layout(height=380).show()
        else:
            display(HTML("<b>No gender data to display.</b>"))

        # Sector x gender stacked
        pivot = df.pivot_table(index='sector', columns='gender_bucket', values='funded', aggfunc='mean').fillna(0)
        if not pivot.empty:
            for g in ['female','male','mixed','unknown']:
                if g not in pivot.columns: pivot[g] = 0
            px.bar(pivot, x=pivot.index, y=['female','male','mixed','unknown'], 
                   title='Funding Rate by Sector & Gender', text_auto='.2f').update_layout(height=400, xaxis_tickangle=-45).show()
        else:
            display(HTML("<b>No sector x gender data to display.</b>"))

        # Prediction sample
        sample = df.sample(min(5000, len(df)), random_state=42).reset_index(drop=True)
        if sample.empty:
            display(HTML("<b>No data for prediction plots.</b>"))
            return

        cat_cols = ['country','region','sector','activity','partner_id','gender_bucket']
        num_cols = ['loan_amount','term_in_months','lender_count']
        X_num = sample[num_cols].values

        # Ensure all categorical columns have categories expected by OHE
        for i, col in enumerate(cat_cols):
            sample[col] = pd.Categorical(sample[col], categories=models['ohe'].categories_[i])

        X_cat = models['ohe'].transform(sample[cat_cols])
        try:
            X_cat = X_cat.toarray()
        except:
            pass

        X_pred = np.hstack([X_num, X_cat])
        sample['predicted_funding_prob'] = models['clf'].predict_proba(X_pred)[:,1]
        sample['predicted_duration'] = models['reg'].predict(X_pred)

        # Predicted funding probability histogram
        if len(sample) > 0:
            px.histogram(sample, x='predicted_funding_prob', nbins=50, title='Predicted Funding Probability')\
                .update_layout(height=360).show()

        # Actual vs predicted funding duration scatter
        if len(sample) > 0:
            fig_sc = px.scatter(sample, x='funding_duration_days', y='predicted_duration', 
                                color='sector', title='Actual vs Predicted Funding Duration', opacity=0.65)
            maxv = max(sample['funding_duration_days'].max(), sample['predicted_duration'].max())
            fig_sc.add_shape(type='line', x0=0, x1=maxv, y0=0, y1=maxv, line=dict(dash='dash', color='red'))
            fig_sc.update_layout(height=500).show()

    widgets.interact(notebook_dashboard,
                     country_filter=widgets.SelectMultiple(options=countries, description='Country'),
                     sector_filter=widgets.SelectMultiple(options=sectors, description='Sector'),
                     gender_filter=widgets.SelectMultiple(options=genders, description='Gender'))
```

##### Bar charts, interactive maps, treemaps, and summary metrics are all produced by the dashboard.  Additionally, it displays anticipated funding durations and probabilities, enabling dynamic data exploration.

_This cell defines a function to generate additional visualizations using data on poverty from the MPI dataset.  A choropleth map of loan counts, a choropleth map of poverty (MPI), a bubble map displaying loan volume versus MPI, and a treemap displaying loan distribution by country and sector are among the interactive visualizations that are created by combining country-level MPI values with the loan data.  Additionally, it conducts a correlation analysis using MPI and a heatmap that illustrates the connections between loan characteristics and poverty levels.  These images shed light on the relationship between loan activity and socioeconomic circumstances in various nations._

```python
#  Additional Visualizations (corrected for your MPI file)
def mpi_visualisation(loans, data_path=DATA):
    # Load MPI data for poverty visuals
    mpi = pd.read_csv(data_path / 'kiva_mpi_region_locations.csv')

    # Merge with loans for country-level stats
    country_stats = loans.groupby('country', observed=True).agg(
        num_loans=('id','count'),
        total_amount=('loan_amount','sum'),
        funding_rate=('funded','mean')
    ).reset_index()
    country_mpi = mpi[['country','MPI']].drop_duplicates()
    country_merge = country_stats.merge(country_mpi, on='country', how='left')

    # --- World Map: Number of Loans ---
    fig1 = px.choropleth(
        country_merge, locations='country', locationmode='country names',
        color='num_loans', hover_data=['total_amount','funding_rate','MPI'],
        title='Global Kiva Loans'
    )
    fig1.update_layout(height=600)
    fig1.show()

    # --- World Map: Poverty (MPI) ---
    fig2 = px.choropleth(
        country_merge, locations='country', locationmode='country names',
        color='MPI', hover_data=['num_loans','total_amount','funding_rate'],
        title='Global Poverty (MPI)'
    )
    fig2.update_layout(height=600)
    fig2.show()

    # --- Bubble Map: Volume vs MPI ---
    fig3 = px.scatter_geo(
        country_merge, locations='country', locationmode='country names',
        size='total_amount', color='MPI',
        hover_name='country', hover_data=['num_loans','funding_rate'],
        projection='natural earth', title='Loan Volume vs Poverty'
    )
    fig3.update_layout(height=600)
    fig3.show()

    # --- Treemap: Country -> Sector ---
    sector_country = loans.groupby(['country','sector'], observed=True).agg(
        num_loans=('id','count'),
        total_amount=('loan_amount','sum')
    ).reset_index()
    fig4 = px.treemap(
        sector_country, path=['country','sector'], values='num_loans',
        color='total_amount', color_continuous_scale='Viridis',
        title='Loans by Country and Sector'
    )
    fig4.update_layout(height=600)
    fig4.show()

    # --- MPI Correlation Analysis ---
    merged_corr = loans.merge(country_mpi, on='country', how='left')
    num_cols = ['loan_amount','funded_amount','term_in_months','lender_count','funding_duration_days','MPI']
    sample_corr = merged_corr[num_cols].sample(n=min(20000,len(merged_corr)), random_state=42)
    corr_matrix = sample_corr.corr()
    plt.figure(figsize=(8,6))
    sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm')
    plt.title('Correlation Matrix including MPI')
    plt.tight_layout()
    plt.show()

# Usage
mpi_visualisation(loans)
```

##### Global loan distributions, poverty (MPI) levels, loan volume versus poverty, country-sector treemaps, and correlation heatmaps are all shown in the prior output. These illustrations show how loans and socioeconomic circumstances are related.

#### _Loan Count by MPI Category_
